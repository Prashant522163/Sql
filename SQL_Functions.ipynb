Q1. Create a ranking of students based on score (highest first).
Function used: RANK()
SELECT
    name,
    score,
    RANK() OVER (ORDER BY score DESC) AS Student_Rank
FROM Student_Performance;


Explanation:
- RANK() assigns a rank to each student.
- ORDER BY score DESC sorts students from highest score to lowest score.
- The student with the highest score gets rank 1.
Expected top results: Arjun Verma (95) – Rank 1, Ishita Joshi (93) – Rank 2, Priya Desai (92) – Rank 3.

Q2. Show each student's score and the previous student's score (based on score order).
Function used: LAG()
SELECT
    name,
    score,
    LAG(score, 1, 0) OVER (ORDER BY score DESC) AS Previous_Score
FROM Student_Performance;


Explanation:
- LAG() retrieves the value from the previous row.
- 1 means one row before.
- 0 is the default value if there is no previous row.
- ORDER BY score DESC sorts students by highest score first.
Example: Arjun – 95 – Previous Score: 0; Ishita – 93 – Previous Score: 95.

Q4. Show each student's name and the next student's attendance (ordered by attendance).
Function used: LEAD()
SELECT
    name,
    attendance,
    LEAD(attendance, 1, 0) OVER (
        ORDER BY attendance
    ) AS Next_Attendance
FROM Student_Performance;


Explanation:
- LEAD() retrieves the value from the next row.
- 1 means the next row.
- 0 is returned when there is no next row.
- ORDER BY attendance sorts attendance in ascending order.
Example: Ayesha Khan – 81 – Next Attendance: 82.

Q5. Assign students into 4 performance groups using NTILE().
Function used: NTILE(4)
SELECT
    name,
    score,
    NTILE(4) OVER (ORDER BY score DESC) AS Performance_Group
FROM Student_Performance;


Explanation:
- NTILE(4) divides students into 4 approximately equal groups.
- ORDER BY score DESC places students with higher scores first.
- Each student receives a group number from 1 to 4.
Expected output: 4 performance groups (Group 1, 2, 3 and 4).

Q6. For each course, assign a row number based on attendance (highest first).
Function used: ROW_NUMBER()
SELECT
    name,
    course,
    attendance,
    ROW_NUMBER() OVER (
        PARTITION BY course
        ORDER BY attendance DESC
    ) AS Row_Num
FROM Student_Performance;


Explanation:
- ROW_NUMBER() assigns a unique row number.
- PARTITION BY course creates separate groups for each course.
- ORDER BY attendance DESC ranks students according to highest attendance first.
Example: Python – Kabir Khanna (96) gets Row Number 1.

Q7. Calculate the number of days each student has been enrolled (from join_date to today).
Function used: DATEDIFF()

The question assumes the current date is 2025-01-01.
SELECT
    name,
    join_date,
    DATEDIFF('2025-01-01', join_date) AS Days_Enrolled
FROM Student_Performance;


Explanation:
- DATEDIFF() calculates the difference between two dates in days.
- First date = 2025-01-01.
- Second date = join_date.
- The result shows the number of days each student has been enrolled.
Example: Aarav Mehta joined on 2023-06-12 – 569 days enrolled.

Q10. For each course, find the highest score using FIRST_VALUE().
Function used: FIRST_VALUE()
SELECT
    name,
    course,
    score,
    FIRST_VALUE(score) OVER (
        PARTITION BY course
        ORDER BY score DESC
    ) AS Highest_Score
FROM Student_Performance;


Explanation:
- FIRST_VALUE() returns the first value in the ordered window.
- PARTITION BY course creates separate groups for each course.
- ORDER BY score DESC places the highest score first.
- Every student in the same course gets that course's highest score.
Expected highest scores:
Course	Highest Score	Student
Data Science	93	Ishita Joshi
Python	91	Kabir Khanna
SQL	92	Priya Desai
Tableau	95	Arjun Verma
