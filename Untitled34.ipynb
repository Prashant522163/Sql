Q1. Create a New Database and Employees Table
Step 1: Create database
CREATE DATABASE company_db;

After creating the database, connect to company_db in pgAdmin.

Step 2: Create Employees table
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    department VARCHAR(50),
    salary INT,
    hire_date DATE
);

Q3. Display All Employee Records Sorted by Salary

Lowest to highest means ASC.

SELECT *
FROM employees
ORDER BY salary ASC;
Remember:
ASC  -- Lowest to Highest
DESC -- Highest to Lowest
Q4. Show Employees Sorted by Department A–Z and Salary High → Low
SELECT *
FROM employees
ORDER BY department ASC, salary DESC;

This means:

Sort department from A to Z
Within each department, sort salary from highest to lowest

Q9. Difference Between Primary Key and Foreign Key
Primary Key	Foreign Key
Uniquely identifies each row	Connects two tables
Cannot have duplicate values	Duplicate values are allowed
Cannot normally be NULL	Can be NULL unless NOT NULL is specified
Used to identify records	Used to create relationships
Example: employee_id	Example: department_id
Example
employee_id INT PRIMARY KEY

Here, employee_id uniquely identifies an employee.

A foreign key example:

department_id INT,
FOREIGN KEY (department_id)
REFERENCES departments(department_id)
Easy way to remember:

Primary Key → identifies the record

Foreign Key → connects the records/tables

Q10. What Are Constraints in SQL?

Constraints are rules applied to columns in a table to maintain data accuracy and integrity.

Common SQL constraints
Constraint	Purpose
PRIMARY KEY	Uniquely identifies each row
FOREIGN KEY	Connects two tables
NOT NULL	Does not allow empty/NULL values
UNIQUE	Does not allow duplicate values
CHECK	Ensures a condition is satisfied
DEFAULT	Provides a default value
Example
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    salary INT CHECK (salary > 0),
    email VARCHAR(100) UNIQUE
);

Here:

PRIMARY KEY → employee ID must be unique
NOT NULL → first name cannot be empty
CHECK → salary must be greater than 0
UNIQUE → email cannot be duplicated
