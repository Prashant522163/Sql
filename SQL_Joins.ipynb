Question 1. Retrieve all customers who have placed at least one order.
Join Used: INNER JOIN
SELECT DISTINCT
    C.CustomerID,
    C.CustomerName,
    C.City
FROM Customers C
INNER JOIN Orders O
ON C.CustomerID = O.CustomerID;


Explanation: INNER JOIN returns only customers whose CustomerID matches in both tables.
Question 2. Retrieve all customers and their orders, including customers who have not placed any orders.
Join Used: LEFT JOIN
SELECT
    C.CustomerID,
    C.CustomerName,
    O.OrderID,
    O.OrderDate,
    O.Amount
FROM Customers C
LEFT JOIN Orders O
ON C.CustomerID = O.CustomerID;


Explanation: LEFT JOIN returns all customers, even if they have no orders. Their order details will appear as NULL.
Question 3. Retrieve all orders and their corresponding customers, including orders placed by unknown customers.
Join Used: RIGHT JOIN
SELECT
    O.OrderID,
    O.CustomerID,
    O.OrderDate,
    O.Amount,
    C.CustomerName,
    C.City
FROM Customers C
RIGHT JOIN Orders O
ON C.CustomerID = O.CustomerID;


Explanation: RIGHT JOIN returns all orders, including orders whose CustomerID does not exist in the Customers table.
Question 4. Display all customers and orders, whether matched or not.
Join Used: FULL OUTER JOIN (using UNION)

MySQL does not directly support FULL OUTER JOIN, so we use UNION.
SELECT
    C.CustomerID,
    C.CustomerName,
    O.OrderID,
    O.OrderDate,
    O.Amount
FROM Customers C
LEFT JOIN Orders O
ON C.CustomerID = O.CustomerID

UNION

SELECT
    C.CustomerID,
    C.CustomerName,
    O.OrderID,
    O.OrderDate,
    O.Amount
FROM Customers C
RIGHT JOIN Orders O
ON C.CustomerID = O.CustomerID;


Explanation: This query returns all customers and all orders, whether they match or not.
Question 5. Find customers who have not placed any orders.
Join Used: LEFT JOIN + IS NULL
SELECT
    C.CustomerID,
    C.CustomerName,
    C.City
FROM Customers C
LEFT JOIN Orders O
ON C.CustomerID = O.CustomerID
WHERE O.OrderID IS NULL;


Explanation: This query returns only those customers who do not have any matching order.
Question 6. Retrieve customers who made payments but did not place any orders.
Join Used: INNER JOIN + LEFT JOIN
SELECT
    C.CustomerID,
    C.CustomerName,
    P.PaymentID,
    P.Amount AS PaymentAmount
FROM Customers C
INNER JOIN Payments P
ON C.CustomerID = P.CustomerID
LEFT JOIN Orders O
ON C.CustomerID = O.CustomerID
WHERE O.OrderID IS NULL;


Explanation: First, INNER JOIN finds customers who made payments. Then LEFT JOIN with IS NULL excludes customers who placed orders.
Question 7. Generate a list of all possible combinations between Customers and Orders.
Join Used: CROSS JOIN
SELECT
    C.CustomerID,
    C.CustomerName,
    O.OrderID,
    O.Amount
FROM Customers C
CROSS JOIN Orders O;


Explanation: CROSS JOIN generates every possible combination of customers and orders.

Since there are 5 customers and 5 orders:

Total combinations = 5 × 5 = 25 rows.
Question 8. Show all customers along with order and payment amounts in one table.
Join Used: LEFT JOIN
SELECT
    C.CustomerID,
    C.CustomerName,
    O.OrderID,
    O.Amount AS OrderAmount,
    P.PaymentID,
    P.Amount AS PaymentAmount
FROM Customers C
LEFT JOIN Orders O
ON C.CustomerID = O.CustomerID
LEFT JOIN Payments P
ON C.CustomerID = P.CustomerID;


Explanation: This query displays customers along with their order and payment information. NULL appears where information is unavailable.
Question 9. Retrieve all customers who have both placed orders and made payments.
Join Used: INNER JOIN
SELECT DISTINCT
    C.CustomerID,
    C.CustomerName,
    C.City
FROM Customers C
INNER JOIN Orders O
ON C.CustomerID = O.CustomerID
INNER JOIN Payments P
ON C.CustomerID = P.CustomerID;


Explanation: INNER JOIN with both Orders and Payments returns customers who have at least one order and at least one payment.

4. Quick Revision – Types of SQL Joins
INNER JOIN
Matching records

Returns only matching rows from both tables.

LEFT JOIN

Returns all rows from the left table and matching rows from the right table.

RIGHT JOIN

Returns all rows from the right table and matching rows from the left table.

FULL OUTER JOIN

Returns all rows from both tables, matching where possible.

CROSS JOIN

Returns every possible combination of rows from both tables.

5 customers × 5 orders = 25 combinations

5. Important: Self JOIN using Employees
Since you also have an Employees table, you can practice SELF JOIN to find each employee's manager.
SELECT
    E.EmployeeName AS Employee,
    M.EmployeeName AS Manager
FROM Employees E
LEFT JOIN Employees M
ON E.ManagerID = M.EmployeeID;


Explanation: The Employees table joins with itself to display employee and manager names.
