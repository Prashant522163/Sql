Q1. What is a Common Table Expression (CTE), and how does it improve SQL query readability?
Answer:

A Common Table Expression (CTE) is a temporary named result set used within an SQL query. It is created using the WITH keyword.

How it improves readability:
1. Makes complex queries easier to understand.
2. Organizes SQL code into smaller parts.
3. Avoids repeating the same subquery.
4. Supports recursive queries.
Example:
WITH HighSalary AS (
    SELECT *
    FROM Employees
    WHERE Salary > 50000
)
SELECT * FROM HighSalary;


Conclusion: CTE makes complex SQL queries simpler, cleaner, and easier to manage.

Q2. Why are some views updatable while others are read-only? Explain with an example.
Answer:

A view is a virtual table created using an SQL query.

Some views are updatable because they are based on a single table without complex operations. Other views are read-only because they contain operations such as GROUP BY, DISTINCT, or aggregate functions.

Example of an updatable view:
CREATE VIEW EmployeeView AS
SELECT EmployeeID, Name, Salary
FROM Employees;

UPDATE EmployeeView
SET Salary = 60000
WHERE EmployeeID = 1;


Example of a non-updatable view:
CREATE VIEW DepartmentSummary AS
SELECT Department, AVG(Salary) AS AverageSalary
FROM Employees
GROUP BY Department;


Conclusion: Simple views are generally updatable, whereas views containing grouping and aggregate functions are generally read-only.

Q3. What advantages do stored procedures offer compared to writing raw SQL queries repeatedly?
Answer:

A stored procedure is a group of SQL statements stored in a database and executed whenever required.

Advantages:
1. Code Reusability: Write once and execute multiple times.
2. Performance: Reduces repeated query-processing overhead.
3. Security: Controls access to database operations.
4. Less Network Traffic: Multiple SQL statements can execute through one procedure call.
5. Easy Maintenance: SQL logic can be maintained in one place.
Example:
CREATE PROCEDURE GetEmployees()
BEGIN
    SELECT * FROM Employees;
END;


To execute:
CALL GetEmployees();


Conclusion: Stored procedures improve code reusability, security, and database efficiency.

Q4. What is the purpose of triggers in a database? Mention one use case where a trigger is essential.
Answer:

A trigger is a special database object that executes automatically when a specific database event occurs.

Triggers are commonly associated with:
- INSERT
- UPDATE
- DELETE
Use case: Automatic logging of deleted records.

For example, whenever a product is deleted from the Products table, a trigger can automatically save its details in an archive table.

Conclusion: Triggers help automate database operations and maintain data consistency.

Q5. Explain the need for data modelling and normalization when designing a database.
Answer:

Data Modelling: It is the process of designing the structure of a database, including tables, columns, relationships, and constraints.

Normalization: It is the process of organizing data into tables to reduce data duplication and improve data consistency.

Need for data modelling and normalization:
1. Reduces data redundancy.
2. Prevents insertion, deletion, and update anomalies.
3. Improves data consistency.
4. Maintains data integrity.
5. Makes database management easier.
Conclusion: Data modelling creates a proper database structure, while normalization reduces duplication and improves data integrity.

Q6. Write a CTE to calculate the total revenue for each product and return only products where revenue is greater than 3000.
Answer:

Revenue is calculated using:
\[
Revenue = Price \times Quantity
\]

SQL Query:
WITH ProductRevenue AS (
    SELECT
        p.ProductID,
        p.ProductName,
        SUM(p.Price * s.Quantity) AS TotalRevenue
    FROM Products p
    JOIN Sales s
        ON p.ProductID = s.ProductID
    GROUP BY p.ProductID, p.ProductName
)
SELECT *
FROM ProductRevenue
WHERE TotalRevenue > 3000;


Expected Output:
ProductID	ProductName	TotalRevenue
1	Keyboard	4800
2	Mouse	8000
3	Chair	5000
4	Desk	5500

Explanation:
- The CTE calculates the revenue of each product.
- JOIN connects Products and Sales tables.
- SUM() calculates total revenue.
- WHERE returns only products with revenue greater than 3000.

Q7. Create a view named vw_CategorySummary that shows Category, TotalProducts, and AveragePrice.
Answer:

A view is a virtual table that displays the result of an SQL query.

SQL Query:
CREATE VIEW vw_CategorySummary AS
SELECT
    Category,
    COUNT(ProductID) AS TotalProducts,
    AVG(Price) AS AveragePrice
FROM Products
GROUP BY Category;


To display the view:
SELECT * FROM vw_CategorySummary;


Expected Output:
Category	TotalProducts	AveragePrice
Electronics	2	1000
Furniture	2	4000

Explanation:
- COUNT() counts the products in each category.
- AVG() calculates the average price.
- GROUP BY groups products according to category.

Q8. Create an updatable view containing ProductID, ProductName, and Price. Then update the price of ProductID = 1 using the view.
Answer:

An updatable view allows us to modify the original table's data through the view.

Step 1: Create the view
CREATE VIEW vw_ProductDetails AS
SELECT ProductID, ProductName, Price
FROM Products;


Step 2: Update the price
UPDATE vw_ProductDetails
SET Price = 1500
WHERE ProductID = 1;


Step 3: Display updated records
SELECT * FROM vw_ProductDetails;


Explanation:
- The view displays ProductID, ProductName, and Price.
- The UPDATE statement changes the price of ProductID 1.
- Since this is a simple view based on one table, the change is reflected in the original Products table.

Q9. Create a stored procedure that accepts a category name and returns all products belonging to that category.
Answer:

A stored procedure is a pre-written SQL program that can accept parameters and execute queries.

SQL Query:
DELIMITER //

CREATE PROCEDURE GetProductsByCategory(
    IN cat_name VARCHAR(50)
)
BEGIN
    SELECT *
    FROM Products
    WHERE Category = cat_name;
END //

DELIMITER ;


To execute the procedure:
CALL GetProductsByCategory('Electronics');


Expected Output:
ProductID	ProductName	Category	Price
1	Keyboard	Electronics	1200
2	Mouse	Electronics	800

Explanation:
- CREATE PROCEDURE creates a stored procedure.
- IN accepts the category name as input.
- WHERE filters products based on the given category.
- CALL executes the procedure.

Q10. Create an AFTER DELETE trigger on the Products table that archives deleted product rows into a new table ProductArchive.
The archive should store ProductID, ProductName, Category, Price, and DeletedAt timestamp.

Answer:

A trigger automatically executes when a specified database event occurs.

Step 1: Create the archive table
CREATE TABLE ProductArchive (
    ProductID INT,
    ProductName VARCHAR(100),
    Category VARCHAR(50),
    Price DECIMAL(10,2),
    DeletedAt TIMESTAMP
);


Step 2: Create the AFTER DELETE trigger
DELIMITER //

CREATE TRIGGER AfterProductDelete
AFTER DELETE ON Products
FOR EACH ROW
BEGIN
    INSERT INTO ProductArchive
    (
        ProductID,
        ProductName,
        Category,
        Price,
        DeletedAt
    )
    VALUES
    (
        OLD.ProductID,
        OLD.ProductName,
        OLD.Category,
        OLD.Price,
        CURRENT_TIMESTAMP
    );
END //

DELIMITER ;


Step 3: Delete a product to test the trigger
DELETE FROM Products
WHERE ProductID = 1;


Step 4: Check archived records
SELECT * FROM ProductArchive;


Explanation:
- AFTER DELETE executes after a product is deleted.
- OLD refers to the deleted row's previous values.
- INSERT INTO saves the deleted product details in ProductArchive.
- CURRENT_TIMESTAMP stores the deletion date and time.
Conclusion: This trigger automatically maintains a record of deleted products for future reference.
