Q1. Retrieve the names of employees who earn more than the average salary of all employees.
Solution:
SELECT name, salary
FROM Employees
WHERE salary > (
    SELECT AVG(salary)
    FROM Employees
);


Explanation:
- Inner query calculates the average salary of all employees.
- Outer query retrieves employees whose salary is greater than that average.
Expected output: Abhishek, Priya, Rohit, Neha, Sneha, Kiran, Tanuja

Q2. Find the employees who belong to the department with the highest average salary.

Solution:
SELECT name, department_id, salary
FROM Employees
WHERE department_id = (
    SELECT department_id
    FROM Employees
    GROUP BY department_id
    ORDER BY AVG(salary) DESC
    LIMIT 1
);


Explanation:
- Inner query groups employees by department.
- Calculates the average salary of each department.
- LIMIT 1 selects the department with the highest average salary.
- Outer query retrieves employees from that department.
Expected output: Sneha, Kiran

Q13. Retrieve department names where the total sales exceed ₹10,000.
Solution:

SELECT department_name
FROM Departments
WHERE department_id IN (
    SELECT department_id
    FROM Employees
    WHERE emp_id IN (
        SELECT emp_id
        FROM Sales
    )
    GROUP BY department_id
    HAVING SUM(
        (
            SELECT 0
        )
    ) > 10000
);


Note: The above nested aggregate approach is not valid for calculating total sales in MySQL. Use the following correct subquery solution instead:
SELECT department_name
FROM Departments
WHERE department_id IN (
    SELECT E.department_id
    FROM Employees E
    JOIN Sales S
    ON E.emp_id = S.emp_id
    GROUP BY E.department_id
    HAVING SUM(S.sale_amount) > 10000
);


Explanation:
- Employees and Sales are joined using emp_id.
- GROUP BY groups sales department-wise.
- SUM() calculates total sales of each department.
- HAVING selects departments whose total sales exceed ₹10,000.
Expected output: Sales, Marketing, Finance, IT
